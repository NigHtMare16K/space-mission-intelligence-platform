```python
def max_success():
    max_perc = 0
    vehicle = ""

    for col in df['Launch_Vehicle'].unique():

        ari = df[df['Launch_Vehicle'] == col]

        succ = len(ari[ari['Status'] == 'Success'])
        fail = len(ari[ari['Status'] == 'Failed'])

        if succ + fail == 0:
            continue

        perc = (succ / (succ + fail)) * 100

        if perc > max_perc:
            max_perc = perc
            vehicle = col

    return max_perc, vehicle
```
